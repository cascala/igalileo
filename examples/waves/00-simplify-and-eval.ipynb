# Oscillations and waves 0: Simplify and eval

*Learning path: oscillations and waves, start here.* [Index](README.md) · Next: [The harmonic oscillator](01-harmonic-oscillator.ipynb)

Galileo works answers out exactly: $1/3 + 2/6$ is $2/3$, not $0.6666666666666666$. When you want digits, you
ask for them with `eval`. This short lesson shows how answers are written, how to get digits, and the two
settings that shape every answer:

* `calc` decides what Galileo does with what you type. The default is `calc = "simplify"`.
* `explain` decides how much each command explains. The default is `explain = "brief"`.

## 1. Exact answers

Click into a code cell and press **Shift+Enter** to run it; the answer appears underneath. Every cell can be
changed and run again.

```galileo
1/3 + 2/6
sqrt(8)
x = 3;
x + 7
```

Fractions and roots stay exact, and nothing is ever rounded: $1/3$ is one third. A name you give a value, like
`x`, is replaced by that value.

## 2. One fixed form

Galileo writes every answer in one fixed form: fractions added, common factors cancelled, products multiplied
out. Every way of writing the same expression ends in the same form. The command `simplify` gives this form
too, and simplifying twice changes nothing.

```galileo
clear
(y^2 - 1)/(y - 1)
(y + 1)^2
simplify(simplify((y + 1)^2))
factor(y^2 + 2*y + 1)
```

When you want another form, ask for it: `factor` gives the factored form, `expand` the multiplied-out one.

## 3. `eval`: digits

```galileo
eval(1/3 + 2/6)
eval(pi, 30)
sin(pi/5)
eval(sin(pi/5))
```

`eval(x)` gives about 16 significant digits, and `eval(x, n)` gives `n` of them. Functions of exact numbers, like
$\sin(\pi/5)$, stay exact until you call `eval`.

Decimals you type are exact numbers too, so there is no rounding error until you ask for digits:

```galileo
0.1 + 0.2 == 0.3
0.1*y + 0.2*y == 0.3*y
```

## 4. In this path

A formula with numbers put in gives an exact answer.

```galileo
k = 4; m = 0.25;
omega = sqrt(k/m)
eval(2*pi/omega)
```

A 0.25 kg mass on a spring of stiffness 4 N/m swings at $\omega = \sqrt{k/m} = 4$ radians per second, so one swing takes $2\pi/4 \approx 1.57$ s.

## 5. The `calc` setting

`calc` decides what Galileo does with what you type. It has three values:

* `calc = "simplify"`, the default: answers are worked out exactly, as above.
* `calc = "eval"`: answers come out in digits, as on a calculator. Whole numbers stay whole.
* `calc = "off"`: what you type stays as you typed it, so you can check it, until you call `simplify` or `eval`.

```galileo
calc = "eval";
1/3 + 2/6
sqrt(8)
calc = "off";
1/3 + 2/6
simplify(1/3 + 2/6)
eval(1/3 + 2/6)
calc = "simplify";
```

In every mode, commands such as `solve`, `factor` and `diff` work out their answers, and comparisons give `true`
or `false`. `clear` removes your variables but keeps the setting.

## 6. The `explain` setting

Many commands add a short explanation to their answer, set apart from it (in grey, or as lines starting with `//`).
The answer itself is always something you could type back into Galileo. `explain` sets how much is added:

* `explain = "brief"`, the default: one or two sentences.
* `explain = "teach"`: a longer explanation that walks through the method, for learning.
* `explain = "off"`: answers only.

```galileo
clear
solve(x^2 - 5*x + 6 == 0, x)
explain = "teach";
solve(x^2 - 5*x + 6 == 0, x)
explain = "off";
solve(x^2 - 5*x + 6 == 0, x)
explain = "brief";
```

A `;` at the end of a line hides both the answer and its explanation.

## Summary

* Galileo works answers out exactly, in one fixed form; `simplify(x)` gives that form on request.
* `eval(x)` gives digits; `eval(x, n)` gives `n` of them.
* `factor` and `expand` give other forms.
* `calc = "simplify"` (the default) works answers out exactly, `calc = "eval"` gives every answer in digits,
  and `calc = "off"` keeps what you type as typed.
* `explain = "brief"` (the default) adds a short explanation, `explain = "teach"` a longer one, and
  `explain = "off"` none.
