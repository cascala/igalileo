# Foundations 1: Numbers and arithmetic

*Learning path: foundations, part 1 of 9.* [Index](README.md) · Next: [Algebra](02-algebra.ipynb)

This first lesson shows how to type calculations into Galileo and how it treats numbers. You will learn

* how to run a cell and read its answer,
* why Galileo keeps **fractions and roots exact**, and how to get a decimal when you want one,
* powers, roots, very large whole numbers and the order of operations,
* whole-number tools: remainders, greatest common divisors and prime factors,
* how to store results in **variables**.

No background is needed beyond school arithmetic.

## 1. Your first calculation

Click into the cell below and press **Shift+Enter** to run it. Galileo prints the answer underneath. Every code cell
in these lessons can be changed and run again, so experiment freely.

```galileo
2 + 3*4
```

The answer is always something you could type into Galileo yourself. When a command does more than arithmetic, an
**explanation** follows the answer, set apart from it (in grey, or as lines starting with `//`). Try it with the
prime factors of 360:

```galileo
factor(360)
```

`explain = "off"` switches explanations off for the rest of the session, and `explain = "brief"` switches them back on
(`explain = "teach"` gives longer, step-by-step ones).
A `;` at the end of a line hides the answer and its explanation.

Multiplication is written `*` and happens before addition, as in school: $2 + 3 \cdot 4 = 14$. Use brackets to change
the order:

```galileo
(2 + 3)*4
```

The five arithmetic operators are `+`, `-`, `*`, `/` and `^` (power). Several lines in one cell are computed one
after the other, and each answer is shown:

```galileo
7 - 10
2^10
10/4
```

Look at the last answer: $10/4$ came out as $5/2$, not $2.5$. That is the first big idea of this lesson.

## 2. Exact answers

A calculator works with decimals and rounds after every step. Galileo keeps whole numbers and fractions **exact**:
it reduces fractions to lowest terms and never rounds.

```galileo
1/3 + 1/6
1/3 + 1/3 + 1/3
2/3 * 9/4
```

On a calculator $1/3$ becomes $0.3333333$, and three of those add up to $0.9999999$, not $1$. Exact arithmetic
avoids that kind of error completely.

Square roots stay exact too. Galileo simplifies them by taking square factors out of the root:
$\sqrt{12} = \sqrt{4 \cdot 3} = 2\sqrt{3}$.

```galileo
sqrt(12)
sqrt(2)*sqrt(2)
sqrt(50)/sqrt(2)
```

### Getting a decimal

When you want a decimal number, wrap the expression in `eval(...)`. It computes the value with about 16 significant
digits. `pi` and `e` are built-in constants:

```galileo
eval(1/3)
eval(sqrt(2))
eval(pi)
eval(e)
```

Typing a decimal point also asks for a decimal answer. Numbers with a decimal point are *approximate*, and anything
combined with an approximate number becomes approximate:

```galileo
1/3 + 0.5
2.0^10
```

**Rule of thumb:** type whole numbers and fractions when you want exact answers; add `eval(...)` at the end when you
want to know "about how much".

## 3. Powers, roots and big numbers

`a^b` is $a$ to the power $b$. Negative and fractional powers mean the same as in algebra:
$a^{-1} = 1/a$ and $a^{1/2} = \sqrt{a}$, $a^{1/3} = \sqrt[3]{a}$.

```galileo
2^-3
8^(1/3)
16^(3/4)
```

Whole numbers have no size limit, so you can see every digit of large powers and factorials.
`factorial(n)` is $n! = 1 \cdot 2 \cdot 3 \cdots n$, the number of ways to put $n$ things in a row; $52!$ is the
number of ways to shuffle a deck of cards:

```galileo
2^100
factorial(52)
```

### Order of operations

Galileo follows the usual rules: brackets first, then powers, then multiplication and division (left to right), then
addition and subtraction (left to right). A minus sign in front binds more loosely than a power, so $-2^2$ means
$-(2^2) = -4$, just as in algebra:

```galileo
-2^2
(-2)^2
2^3^2
24/4/2
```

Powers are evaluated from the right: $2^{3^2} = 2^9 = 512$. Division goes from the left: $24/4/2 = 6/2 = 3$. When in
doubt, add brackets: they never hurt.

## 4. Whole numbers: remainders, divisors and primes

Dividing $17$ by $5$ gives $3$ with remainder $2$. `mod(a, b)` is the remainder of $a$ divided by $b$, and
`floor(x)` rounds down to a whole number:

```galileo
floor(17/5)
mod(17, 5)
```

The remainder is how clocks work: 10 hours after 9 o'clock it is `mod(9 + 10, 12)` o'clock. Remainders are the start
of the [number theory path](../number-theory/README.md).

```galileo
mod(9 + 10, 12)
```

The **greatest common divisor** `gcd(a, b)` is the largest number that divides both; the **least common multiple**
`lcm(a, b)` is the smallest number both divide. To add $\frac{1}{12} + \frac{1}{18}$ by hand you need the common
denominator `lcm(12, 18)`, and to simplify $\frac{12}{18}$ you divide by `gcd(12, 18)`:

```galileo
gcd(12, 18)
lcm(12, 18)
1/12 + 1/18
```

Every whole number above 1 is a product of **primes** (numbers with no divisors except 1 and themselves) in exactly one
way. `factor(n)` finds it; `isprime(n)` tests a number; `primes(n)` lists the primes up to $n$:

```galileo
factor(360)
isprime(97)
primes(50)
```

`factor` answers with one row per prime: the prime and how many times it divides. So $360 = 2^3 \cdot 3^2 \cdot 5$,
which the explanation writes out. From the factorization you can read off the gcd: $12 = 2^2\cdot 3$ and
$18 = 2 \cdot 3^2$ share $2 \cdot 3 = 6$.

## 5. Variables

A name followed by `=` stores a value. Afterwards the name can be used in calculations. A line that ends with `;` is
computed but its answer is not shown, which is useful for setting values:

```galileo
price = 24;
tax = 8/100;
price*(1 + tax)
```

The answer is exact: $648/25$. As a decimal that is

```galileo
eval(price*(1 + tax))
```

Names can contain letters, digits and `_`, and they are case sensitive (`Price` and `price` are different). If you use
a name that has no value, Galileo keeps it as a symbol. That is the subject of the next lesson:

```galileo
total = price*(1 + rate)
```

Here `price` was replaced by 24, but `rate` has no value, so it stays a letter. Give it one and ask again:

```galileo
rate = 1/4;
total
```

`who` lists the variables you have defined, and `clear` removes them all:

```galileo
who
clear
```

## Summary

* Galileo computes **exactly** with whole numbers, fractions and roots; `eval(...)` or a decimal point gives decimals.
* Operators: `+ - * / ^`; the usual order of operations; brackets to be sure.
* `mod`, `gcd`, `lcm`, `factor`, `isprime` and `primes` work with whole numbers.
* `name = value` stores a value; `;` hides an answer; unknown names stay as symbols.

**Try it:** what is $2^{64}$? Which is larger, $2^{100}$ or $100!$? Is $2^{31} - 1$ prime?
